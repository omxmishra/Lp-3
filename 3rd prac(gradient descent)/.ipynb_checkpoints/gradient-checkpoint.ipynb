```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
def function(x):
    return (x + 3) ** 2

def derivative(x):
    return 2 * (x + 3)
```

```python
x = 2
learning_rate = 0.1
iterations = 100

for i in range(iterations):
    x = x - learning_rate * derivative(x)

print("Local Minimum x:", x)
print("Minimum Value y:", function(x))
```

```python
x_values = np.linspace(-6, 4, 100)
y_values = function(x_values)

plt.plot(x_values, y_values)
plt.scatter(x, function(x), color='red')
plt.xlabel("x")
plt.ylabel("y")
plt.title("Gradient Descent - Local Minimum")
plt.grid()
plt.show()
```

```python
x = 2
learning_rate = 0.1
iterations = 10

history = []

for i in range(iterations):
    history.append([i, x, function(x)])
    x = x - learning_rate * derivative(x)

history = pd.DataFrame(history, columns=['Iteration', 'X', 'Y'])
print(history)
```
