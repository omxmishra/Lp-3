```python
recursive_count = 0
iterative_count = 0

def fibonacci_recursive(n):
    global recursive_count
    recursive_count += 1

    if n <= 1:
        return n

    return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)


def fibonacci_non_recursive(n):
    global iterative_count

    a, b = 0, 1
    result = []

    for i in range(n):
        result.append(a)
        iterative_count += 1
        a, b = b, a + b

    return result


n = int(input("Enter number of terms: "))

recursive_result = []

for i in range(n):
    recursive_result.append(fibonacci_recursive(i))

print("\nRecursive Fibonacci:")
print(recursive_result)

print("Recursive Step Count:", recursive_count)

iterative_result = fibonacci_non_recursive(n)

print("\nNon-Recursive Fibonacci:")
print(iterative_result)

print("Non-Recursive Step Count:", iterative_count)

print("\nTime Complexity:")
print("Recursive: O(2^n)")
print("Non-Recursive: O(n)")

print("\nSpace Complexity:")
print("Recursive: O(n)")
print("Non-Recursive: O(1) auxiliary space")
```
