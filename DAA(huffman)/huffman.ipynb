```python

def huffman_encoding(data):
    freq = {}

    for char in data:
        freq[char] = freq.get(char, 0) + 1

    nodes = [[frequency, [char, ""]] for char, frequency in freq.items()]

    while len(nodes) > 1:
        nodes.sort(key=lambda x: x[0])

        left = nodes.pop(0)
        right = nodes.pop(0)

        for pair in left[1:]:
            pair[1] = "0" + pair[1]

        for pair in right[1:]:
            pair[1] = "1" + pair[1]

        nodes.append([left[0] + right[0]] + left[1:] + right[1:])

    codes = dict(nodes[0][1:])

    print("Character Frequencies:")
    print(freq)

    print("\nHuffman Codes:")
    for char, code in codes.items():
        print(char, ":", code)

    encoded = "".join(codes[char] for char in data)

    print("\nEncoded Data:")
    print(encoded)


data = input("Enter a string: ")
huffman_encoding(data)
```
